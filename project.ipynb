## Part 1: Evaluate a Do Nothing Plan

One of the treatment actions is to do nothing.
Calculate the expected discomfort (not rewards) of a policy that always does nothing.

Hint: for this value calculation and later ones, use value iteration.
The analytical solution has difficulties in practice when there is no discount factor.

```python
# =====================
# Important Libraries
# =====================
%pip install numpy
%pip install pandas

import numpy as np
import pandas as pd
```

```python
# ====================================
# Part 1: Evaluate a Do Nothing Plan
# ====================================
# Calcluate expected discomfort (not rewards) of a policy that always does nothing

import numpy as np
import pandas as pd

# Load data
rewards = pd.read_csv("twizzleflu-rewards.tsv", sep="\t")
transitions = pd.read_csv("twizzleflu-transitions.tsv", sep="\t")

action = "do-nothing"
states = rewards["state"].drop_duplicates().tolist()

# For do-nothing policy, convert negative rewards to positive discomfort
discomfort = (
    rewards[rewards["action"] == action]
    .set_index("state")
    .loc[states, "reward"]
    .mul(-1)
    .to_numpy()
)

# Build P[current_state, next_state] for do-nothing action
policy_transitions = transitions[transitions["action"] == action]

P = (
    policy_transitions
    .pivot(index="state", columns="next_state", values="probability")
    .reindex(index=states, columns=states)
    .fillna(0)
    .to_numpy()
)

# Evaluation of fixed policy to compute expected discomfort until recovery
gamma = 1.0
values = np.zeros(len(states))
tolerance = 1e-12

for _ in range(10_000):
    updated_values = discomfort + gamma * (P @ values)
    if np.max(np.abs(updated_values - values)) < tolerance:
        values = updated_values
        break
    values = updated_values
else:
    raise RuntimeError("Policy evaluation did not converge.")
```

Save the expected discomfort by state to a file "do-nothing-discomfort.tsv" with columns state and expected_discomfort.

```python
# ========================================================================
# Save expected discomfort to TSV file named "do-nothing-discomfort.tsv"
# ========================================================================
expected_discomfort = pd.DataFrame({
    "state": states,
    "expected_discomfort": values,
    })

print(expected_discomfort)

# Save file
expected_discomfort.to_csv("do-nothing-discomfort.tsv", sep="\t", index=False)
```

## Part 2: Compute an Optimal Treatment Plan

Compute an optimal treatment plan for Twizzleflu.
It should minimize the expected discomfort (maximize the rewards).

```python
# ===========================================
# Part 2: Compute an Optimal Treatment Plan
# ===========================================
# It should minimize expected discomfort (maximize rewards) 

# Load data
rewards = pd.read_csv("twizzleflu-rewards.tsv", sep="\t")
transitions = pd.read_csv("twizzleflu-transitions.tsv", sep="\t")

actions = rewards["action"].drop_duplicates().tolist()
states = rewards["state"].drop_duplicates().tolist()

# Build reward and transition matrices for all actions and states
R = np.array([
    rewards[rewards["action"] == a].set_index("state").loc[states, "reward"].to_numpy()
    for a in actions
    ])

P = np.array([
    transitions[transitions["action"] == a]
    .pivot(index="state", columns="next_state", values="probability")
    .reindex(index=states, columns=states)
    .fillna(0)
    .to_numpy()
    for a in actions
    ])

gamma = 1.0
v = np.zeros(len(states))

for _ in range(10_000):
    v_new = (R + gamma * P @ v).max(axis=0)
    if np.max(np.abs(v_new - v)) < 1e-12:
        v = v_new
        break
    v = v_new
else:
    raise RuntimeError("Value iteration did not converge.")

q = R + gamma * P @ v
best_actions = [actions[i] for i in q.argmax(axis=0)]
```

Save the optimal actions for each state to a file "minimum-discomfort-actions.tsv" with columns state and action.

```python
# ========================================================================
# Save optimal actions to TSV file named "minimum-discomfort-actions.tsv"
# ========================================================================
optimal_actions = pd.DataFrame({
    "state": states,
    "action": best_actions,
    })

print(optimal_actions)

# Save file
optimal_actions.to_csv("minimum-discomfort-actions.tsv", sep="\t", index=False)
```

## Part 3: Expected Discomfort

Using your previous optimal policy, compute the expected discomfort for each state.

```python
# =============================
# Part 3: Expected Discomfort 
# =============================
# Use previous optimal policy to calculate expected discomfort for each state

# Build reward vector and transition matrix for chosen policy
action_index = {a: i for i, a in enumerate(actions)}
policy_index = [action_index[a] for a in best_actions]
state_index = np.arange(len(states))

R_pi = R[policy_index, state_index]        # shape: states
P_pi = P[policy_index, state_index, :]     # shape: states x states

values = np.zeros(len(states))

for _ in range(10_000):
    updated_values = R_pi + gamma * (P_pi @ values)
    if np.max(np.abs(updated_values - values)) < 1e-12:
        values = updated_values
        break
    values = updated_values
else:
    raise RuntimeError("Policy evaluation did not converge.")
```

Save your results in a file "minimum-discomfort-values.tsv" with columns state and expected_discomfort.

```python
# ===================================================================================
# Save expected discomfort values to TSV file named "minimum-discomfort-values.tsv"
# ===================================================================================
expected_discomfort = pd.DataFrame({
    "state": states,
    "expected_discomfort": -values + 0.0,   # +0.0 avoids "-0.0" for recovered
    })

print(expected_discomfort)

# Save file
expected_discomfort.to_csv("minimum-discomfort-values.tsv", sep="\t", index=False)
```

## Part 4: Minimizing Twizzleflu Duration

Modifiy the Markov decision process to minimize the days until the Twizzle flu is over.
To do so, change the reward function to always be -1 if the current state corresponds to being sick (must have symptoms, exposed does not count) and 0 otherwise.
To be clear, the action does not matter for this reward function.


```python
# ========================================
# Part 4: Minimizing Twizzleflu Duration
# ========================================
# Reward is -1 for each day with symptoms; exposed and recovered days do not count.
is_sick = np.array([state.startswith("symptoms-") for state in states])
R_days = np.tile(np.where(is_sick, -1.0, 0.0), (len(actions), 1))  # actions x states

v_days = np.zeros(len(states))

for _ in range(10_000):
    updated_values = (R_days + gamma * P @ v_days).max(axis=0)
    if np.max(np.abs(updated_values - v_days)) < 1e-12:
        v_days = updated_values
        break
    v_days = updated_values
else:
    raise RuntimeError("Value iteration did not converge.")

q_days = R_days + gamma * P @ v_days
speed_actions = [actions[i] for i in q_days.argmax(axis=0)]
expected_sick_days = -v_days + 0.0   # +0.0 avoids "-0.0" for recovered
```

Save your new reward function in a file "duration-rewards.tsv" in the same format as "twizzleflu-rewards.tsv".

```python
# ===================================================================
# Save new reward function to TSV file named "duration-rewards.tsv"
# ===================================================================
duration_rewards = pd.DataFrame({
    "action": np.repeat(actions, len(states)),
    "state": states * len(actions),             
    "reward": R_days.ravel() + 0.0,             
    })

print(duration_rewards)

# Save file
duration_rewards.to_csv("duration-rewards.tsv", sep="\t", index=False)
```
